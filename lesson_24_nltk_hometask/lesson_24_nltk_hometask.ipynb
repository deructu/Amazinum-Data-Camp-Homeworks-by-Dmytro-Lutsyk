# Lesson 24 NLTK Homework

Only the homework questions and the setup needed to run them are left here.

```python
import nltk
import numpy as np
from nltk import FreqDist
from nltk.corpus import gutenberg, words
from nltk.tokenize import sent_tokenize, word_tokenize

for package in [
    "gutenberg",
    "punkt",
    "punkt_tab",
    "words",
    "averaged_perceptron_tagger",
    "averaged_perceptron_tagger_eng",
]:
    nltk.download(package, quiet=True)
```

```python
moby_raw = gutenberg.raw("melville-moby_dick.txt")

moby_tokens = word_tokenize(moby_raw)
moby_sentences = sent_tokenize(moby_raw)
moby_dist = FreqDist(moby_tokens)

print("Tokens:", len(moby_tokens))
print("Sentences:", len(moby_sentences))
```

## Question 1

What is the lexical diversity of the text? Return the ratio of unique tokens to all tokens.

```python
def answer_one():
    return len(set(moby_tokens)) / len(moby_tokens)

answer_one()
```

## Question 2

What percentage of tokens is `whale` or `Whale`?

```python
def answer_two():
    whale_count = moby_tokens.count("whale") + moby_tokens.count("Whale")
    return whale_count / len(moby_tokens) * 100

answer_two()
```

## Question 3

What are the most frequently occurring tokens and their frequencies?

```python
def answer_three():
    return moby_dist.most_common(10)

answer_three()
```

## Question 4

Which tokens have length greater than 5 and frequency greater than 150?

```python
def answer_four():
    return sorted([
        token
        for token, frequency in moby_dist.items()
        if len(token) > 5 and frequency > 150
    ])

answer_four()
```

## Question 5

Find the longest token and its length.

```python
def answer_five():
    longest_word = max(moby_tokens, key=len)
    return longest_word, len(longest_word)

answer_five()
```

## Question 6

Which unique alphabetic words have frequency greater than 2000? Return `(frequency, word)` tuples.

```python
def answer_six():
    return sorted(
        [
            (frequency, token)
            for token, frequency in moby_dist.items()
            if token.isalpha() and frequency > 2000
        ],
        reverse=True,
    )

answer_six()
```

## Question 7

What is the average number of tokens per sentence?

```python
def answer_seven():
    sentence_lengths = [len(word_tokenize(sentence)) for sentence in moby_sentences]
    return float(np.mean(sentence_lengths))

answer_seven()
```

## Question 8

What are the 5 most frequent parts of speech in the text?

```python
def answer_eight():
    tagged_tokens = nltk.pos_tag(moby_tokens)
    tag_distribution = FreqDist(tag for _, tag in tagged_tokens)
    return tag_distribution.most_common(5)

answer_eight()
```

## Question 9

Create a spelling recommender for `cormulent`, `incendenece`, and `validrate` using edit distance.

```python
def answer_nine(default_words=["cormulent", "incendenece", "validrate"]):
    correct_spellings = words.words()
    recommendations = []

    for misspelled_word in default_words:
        same_first_letter = [
            word
            for word in correct_spellings
            if word.startswith(misspelled_word[0])
        ]
        recommendation = min(
            same_first_letter,
            key=lambda word: (
                nltk.edit_distance(misspelled_word, word, transpositions=True),
                word,
            ),
        )
        recommendations.append(recommendation)

    return recommendations

answer_nine()
```
